Le projet est organisé en cinq parties. Respectez cet ordre : chaque partie s’appuie sur la précédente.
# Partie A — Préparation et analyse exploratoire (EDA)
1.	Charger les données, inspecter la structure (types, dimensions, premières lignes).
2.	Nettoyer : traiter les doublons et les valeurs manquantes (justifiez votre stratégie : suppression vs imputation).
3.	Produire des statistiques descriptives (tendance centrale, dispersion) pour les variables clés.
4.	Identifier au moins trois constats intéressants (ex. catégorie la plus rentable, impact du retard sur la satisfaction, ville la plus active).
# Partie B — Visualisation de données
Produisez des visualisations soignées (titres, axes légendés, couleurs lisibles). Vous devez utiliser les trois bibliothèques :
Avec matplotlib
–	Un histogramme de la distribution des montants de commande.
–	Une série temporelle : évolution du chiffre d’affaires par mois.
Avec seaborn
–	Une heatmap des corrélations entre variables numériques.
–	Un boxplot du montant par catégorie de produit (ou de la satisfaction par retard).
Avec Plotly (interactif)
–	Au moins un graphique interactif (ex. carte/bar chart du CA par ville, ou nuage de points montant × délai colorié par satisfaction).
Conseil
Chaque graphique doit raconter quelque chose. Sous chaque visualisation, écrivez une à deux phrases d’interprétation : que montre ce graphique, et en quoi est-ce utile pour TogoShop ?
# Partie C — Régression : prédire le montant d’une commande
5.	Construire X / y (cible = montant_total_fcfa), préparer un pipeline (imputation, encodage, scaling).

8.	Analyser les variables les plus influentes et interpréter.
# Partie D — Classification : prédire la satisfaction client
9.	Cible = satisfait. Vérifier l’équilibre des classes et faire un split stratifié.
10.	Entraîner au moins deux modèles de classification.
11.	Évaluer avec précision, rappel, F1 et ROC-AUC ; afficher la matrice de confusion.
12.	Discuter : quelle métrique privilégier du point de vue de TogoShop, et pourquoi ?
# Partie E — Clustering : segmenter les clients
13.	Sélectionner des variables comportementales pertinentes et les standardiser.
14.	Déterminer le nombre de segments (méthode du coude + silhouette).
